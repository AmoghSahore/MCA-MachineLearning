# KNN Approaches Used

## Basic idea of KNN

KNN stands for **K-Nearest Neighbours**. When a new tumour record is given, KNN measures its distance from the training records and selects the nearest K records. The class found most often among these neighbours becomes the prediction.

For example, if `K = 5` and three neighbours are malignant while two are benign, the new record is predicted as malignant.

KNN is called a **lazy learning algorithm** because it does not create a mathematical model during training. It stores the training data and performs the main calculations when a prediction is requested.

## 1. KNN classification

The target has two classes: `0` means malignant and `1` means benign. Therefore, this is a **binary classification** problem. KNN predicts the class using majority voting among the nearest neighbours.

## 2. Standardized KNN

The dataset features have different numerical ranges. A feature with large values could dominate the distance calculation. `StandardScaler` converts all features to a common scale before KNN is applied.

**How to explain it:** Scaling is necessary because KNN is distance-based. Without scaling, large-valued features may affect the prediction more than other features.

## 3. Euclidean-distance KNN

Euclidean distance is the normal straight-line distance between two points. It is the default distance method used by KNN and works well with numerical, scaled features.

## 4. Manhattan-distance KNN

Manhattan distance adds the absolute differences between feature values. It is similar to travelling through city blocks instead of moving directly in a straight line. Its performance is compared with Euclidean distance to see which method suits the data better.

## 5. Heuristic K selection

The square root of the number of training samples is used as the initial K value. An odd value is preferred to reduce the possibility of a voting tie. Nearby K values are also tested because the square-root rule only gives a starting estimate.

## 6. Cross-validated KNN

A single train-test split may produce a lucky or unlucky result. In 5-fold cross-validation, the training data is divided into five parts. The model trains on four parts and validates on the remaining part. This is repeated five times, and the average accuracy is calculated. The K with the highest mean accuracy is selected.

**How to explain it:** Cross-validation gives a more reliable K because every K is evaluated on several parts of the data instead of only one split.

## 7. Decision-boundary comparison

Decision boundaries are drawn for `K = 1, 5, 10, and 20` using two features. A small K creates a detailed and irregular boundary and may overfit. A large K creates a smoother boundary but may underfit. A moderate K usually gives a better balance.

## Evaluation metrics

- **Accuracy:** The percentage of all predictions that are correct.
- **Precision:** Out of the records predicted as malignant, how many are actually malignant.
- **Recall:** Out of all actual malignant records, how many are correctly detected.
- **F1 score:** The balance between precision and recall.
- **Confusion matrix:** A table showing correct and incorrect predictions for both classes.
- **ROC-AUC:** Measures how well the model separates malignant and benign records at different thresholds.

Recall for the malignant class is especially important because a false negative means that a malignant case was incorrectly predicted as benign.

## Important terms

- **K:** The number of nearest neighbours checked.
- **Lazy learning:** Storing training data and calculating only when prediction is required.
- **Distance metric:** A method used to measure how close two records are.
- **Majority voting:** Selecting the class that appears most among the K neighbours.
- **Heuristic:** A simple rule used to find a reasonable starting value.
- **Cross-validation:** Repeatedly testing a model on different parts of the training data.
- **Decision boundary:** The border separating regions predicted as different classes.
- **Overfitting:** Learning the training data too closely and performing poorly on new data.
- **Underfitting:** Using a model that is too simple to learn important patterns.
