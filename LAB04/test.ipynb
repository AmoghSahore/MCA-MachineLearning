# Explanation of Graphs

## 1. Accuracy for nearby K values

This graph shows test accuracy for K values from 16 to 26. The highest accuracy is about **98.25% at K = 17 and K = 18**. Accuracy generally decreases when K becomes larger in this range. This happens because a large K considers more distant neighbours and may smooth out useful local patterns.

This graph is used only to study K values near the square-root heuristic. It does not directly decide the final K because the result is based on a single train-test split.

**Terms:**
- **K:** Number of nearest neighbours used for prediction.
- **Accuracy:** Proportion of all test predictions that are correct.
- **Heuristic K:** A starting K estimated using the square root of the training size.

## 2. Decision boundaries for K = 1, 5, 10 and 20

These graphs use only `radius_mean` and `texture_mean` so that the predictions can be shown in two dimensions. The dots are actual records and the background colour is the class predicted by KNN. Blue represents malignant and red represents benign in these plots.

- At **K = 1**, the boundary is irregular and contains small isolated regions. The model follows individual data points closely and may overfit.
- At **K = 5**, the boundary becomes smoother while still keeping local patterns.
- At **K = 10**, the boundary is smoother and less affected by individual records.
- At **K = 20**, the boundary is very smooth because many neighbours take part in each prediction. Too much smoothing can cause underfitting.

These plots explain the effect of K, but they do not show the exact final model boundary because the final model uses all 30 features.

**Terms:**
- **Decision boundary:** The border between regions predicted as different classes.
- **Overfitting:** Following the training data too closely, including small variations.
- **Underfitting:** Using a model that is too simple to learn important patterns.

## 3. Cross-validation accuracy for K values

This graph shows the mean accuracy from 5-fold stratified cross-validation for odd K values from 1 to 39. The highest mean accuracy is about **96.48% at K = 9**, so K = 9 is selected for the final model.

The score is lower at K = 1 because the model is sensitive to individual records. Accuracy also decreases for large K values because the model becomes too smooth. Cross-validation is preferred over the nearby-K graph for final selection because it tests every K on several data sections.

**Terms:**
- **Cross-validation:** Repeatedly training and validating the model on different data sections.
- **Fold:** One section of data used for validation during cross-validation.
- **Mean CV accuracy:** Average validation accuracy across all folds.

## 4. Confusion matrix

The confusion matrix gives the exact number of correct and incorrect predictions made by the final KNN model.

- **39 malignant cases** were correctly predicted as malignant.
- **3 malignant cases** were incorrectly predicted as benign. These are false negatives.
- **0 benign cases** were incorrectly predicted as malignant. These are false positives.
- **72 benign cases** were correctly predicted as benign.

The result shows that the model made only three mistakes, but all three were missed malignant cases. This is important in medical classification because a false negative may delay diagnosis. It also explains why malignant precision is 1.0 while malignant recall is about 0.929.

**Terms:**
- **True positive:** A malignant case correctly predicted as malignant.
- **False negative:** A malignant case incorrectly predicted as benign.
- **False positive:** A benign case incorrectly predicted as malignant.
- **True negative:** A benign case correctly predicted as benign.

## 5. ROC curve for the malignant class

The ROC curve is close to the top-left corner, showing that the final model separates malignant and benign cases very well. The **AUC is 0.994**, which is close to the maximum value of 1.0. The diagonal line represents random classification, while the model's curve stays well above it.

The curve also shows the trade-off between detecting more malignant cases and incorrectly marking benign cases as malignant when the probability threshold changes.

**Terms:**
- **True Positive Rate:** Proportion of malignant cases correctly detected; also called recall.
- **False Positive Rate:** Proportion of benign cases incorrectly predicted as malignant.
- **ROC curve:** True Positive Rate plotted against False Positive Rate at different thresholds.
- **AUC:** Area under the ROC curve; a value closer to 1 indicates better class separation.

## Overall interpretation

The graphs show that K should not be too small or too large. Cross-validation selected **K = 9** as the most reliable value. The final model achieved about **97.37% test accuracy** and **0.994 ROC-AUC**, showing strong performance on this dataset.
